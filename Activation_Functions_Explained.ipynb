```python
def binary_step(x):
    if x<0:
        return 0
    else:
        return 1
print("Activation value for 5 is ", binary_step(5))
print("Activation value for -1 is",binary_step(-1))
print("Activation value for -10 is",binary_step(-1))
print("Activation value for 50 is ", binary_step(5))
```

```python
def linear_function(x):
    return 4*x
print("Linear Function for 4 is:: ",linear_function(4), "\nLinear function for -2 is:: ",linear_function(-2))
```

```python
import numpy as np
def sigmoid_function(x):
    z = (1/(1 + np.exp(-x)))
    return z
print("Sigmoid Function for 4 is:: ",sigmoid_function(4), "\nSigmoid function for -2 is:: ",sigmoid_function(-2))
```

The gradient values are significant for range -3 and 3 but the graph gets much flatter in other regions. This implies that for values greater than 3 or less than -3, will have very small gradients. As the gradient value approaches zero, the network is not really learning.

Also, the sigmoid function is not symmetric around zero. So output of all the neurons will be of the same sign. This can be addressed by scaling the sigmoid function which is exactly what happens in the tanh function

```python
def tanh_function(x):
    z = (2/(1 + np.exp(-2*x))) -1
    return z

print("tanh Function for 4 is:: ",tanh_function(4), "\tanh function for -2 is:: ",tanh_function(-2))
```

the range of values is between -1 to 1

```python
def relu_function(x):
    if x<0:
        return 0
    else:
        return x

print("relu Function for 4 is:: ",relu_function(4), "\n relu function for -4 is:: ",relu_function(-4))
```

gradient of the ReLU function
f'(x) = 1, x>=0
      = 0, x<0
      
      
The Gradient value of the negative side is zero, because of this,  the weights and biases for some neurons are not updated during the backpropogation process.
This is called dead neurons, neurons which never get activated.
This is taken care of by the ‘Leaky’ ReLU function

```python
def leaky_relu_function(x):
    if x<0:
        return 0.01*x
    else:
        return x

print("Leaky relu Function for 4 is:: ",leaky_relu_function(4), "\n leaky relu function for -4 is:: ",leaky_relu_function(-4))
```

Parameterised ReLU - When the value of a is fixed to 0.01, the function acts as a Leaky ReLU function. The parameterized ReLU function is used when the leaky ReLU function still fails to solve the problem of dead neurons and the relevant information is not successfully passed to the next layer.

```python
def elu_function(x, a):
    if x<0:
        return a*(np.exp(x)-1)
    else:
        return x

print("ELU FUnction \n",elu_function(5, 0.1),elu_function(-5, 0.1))
```

    Exp function
    f(x) = x,   x>=0
        = a(e^x-1), x<0
    

    derivative of the elu function for values of  x greater than 0 is 1
    for values of x<0, the derivative would be  a.e^x

    f'(x) = x,   x>=0
        = a(e^x), x<0
    
    

```python
def swish_function(x):
    return x/(1-np.exp(-x))

print(swish_function(-4), swish_function(4))
```

Swish is a lesser known activation function which was discovered by researchers at Google. Swish is as computationally efficient as ReLU and shows better performance than ReLU on deeper models.  The values for swish ranges from negative infinity to infinity



the curve of the function is smooth and the function is differentiable at all points. This is helpful during the model optimization process and is considered to be one of the reasons that swish outoerforms ReLU.

A unique fact about this function is that swich function is not monotonic. This means that the value of the function may decrease even when the input values are increasing.

```python
import numpy as np
def softmax_function(x):
    z = np.exp(x)
    z_ = z/z.sum()
    return z_

softmax_function([0.8, 1.2, 3.1])
```
