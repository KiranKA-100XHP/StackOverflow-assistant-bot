```python
list_of_dishes = ["idli","vada","curd"]
price=[100,200,300]
few_dishes=["vada","idli","kichdi"]
def find_price():
  for each in few_dishes:
    try:
      ind=list_of_dishes.index(each)
      pricex=price[ind]
      print(each,pricex)
    except ValueError as e:
      print(each,e)

find_price()
```
